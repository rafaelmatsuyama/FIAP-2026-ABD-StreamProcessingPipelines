```python
# Dataset Location & Type.
file_location = "/Volumes/workspace/default/teste/Lab 02 - Dataset.json"
file_type = "json"

# JSON Options.
multiline = "true"

# Ingestão dos Dados em formato Batch.
df = spark.read.format(file_type) \
  .option("multiline", multiline) \
  .load(file_location)

# Exibindo os primeiros resultados para verificar visualmente se a carga foi realizada com sucesso.
display(df)
```

```python
# Criando uma View temporária para consultar os dados como um todo.
temp_table_name = "data_json"
df.createOrReplaceTempView(temp_table_name)
```

```python
%sql

/* Consultando a View temporária utilizando SQL. */
select * from `data_json`
```

```python
# Com os dados na base temporária ok, podemos fazer a carga definitiva na base principal.
permanent_table_name = "data_2_json_abdr"
df.write.format("delta").mode("overwrite").saveAsTable(permanent_table_name)
```

```python
%sql

/* Podemos consultar a base definitiva utilizando SQL normalmente. */
select * from `data_2_json_abdr`
```

```python
# Caso prefira fazer as consultas SQL via Python, é possível fazer dessa forma.
result = spark.sql('SELECT * FROM data_2_json_abdr')
result.show()
```
